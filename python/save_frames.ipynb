```python
import numpy as np
import cv2
```

```python
import golf_backend
```

```python
frames = golf_backend.extract_frames("sample.mp4", max_frames=1000)
```

```python
frames.shape
```

```python
from pathlib import Path

img_folder = Path("video_frames")

for index, img in enumerate(frames):
    img_name = img_folder / f"frame_{index:04d}.jpg"
    img_name.parent.mkdir(exist_ok=True, parents=True)
    cv2.imwrite(img_name, img)
```

```python
import matplotlib.pyplot as plt
for frame in frames[:10]:
    plt.imshow(frame)
    plt.axis('off')
    plt.show()
```
